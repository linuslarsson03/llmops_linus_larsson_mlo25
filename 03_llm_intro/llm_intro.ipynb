## Temprature

- High temprature = more creative
- Low temprature = closer to deterministic



```python
from pydantic_ai import ModelSettings

agent = Agent(
"openrouter:nvidia/nemotron-nano-9b-v2:free",
system_prompt="you are a children story teller. Always max 3 sentences.",
model_settings=ModelSettings(temperature=0)
)


prompt = "A goldfish in  a aquarium..."

result = await agent.run(prompt)

print(result.output)
```

```python
agent.model_settings = ModelSettings(temperature=0.5) # more creative
result = await agent.run(prompt)   
print(result.output)
```

```python
# Note can use reasoning tokens if it is a reasoning model, but not in this case since it is a generation model
result.usage()
```

### Multimodal inputs
- text
- audio
- image
- video

```python
from pathlib import Path
from pydantic_ai.messages import BinaryContent

image_agent = Agent(
    "openrouter:qwen/qwen3.6-plus:free",
    system_prompt="You analyze images and text and give back a funny joke about it. Max 2 sentences."
)

image_data = Path("kaninbild.png").read_bytes()

prompt = "What is in this image?"

result = await image_agent.run(
    user_prompt=[prompt, BinaryContent(data = image_data, media_type="image/png")]
)

print(result.output)
```

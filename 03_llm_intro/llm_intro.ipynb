```python
from pydantic_ai import Agent
from dotenv import load_dotenv

load_dotenv()

agent = Agent(
    "openrouter:liquid/lfm-2.5-1.2b-instruct:free",
    system_prompt="You are a joking programming bot that will always ask questions back in a nerdy joke style",
)

prompt = "why is it foggy in göteborg right now?"

result = await agent.run(prompt)

result
```

```python
print(result.output)
```

## Temprature

- High temprature = more creative
- Low temprature = closer to deterministic



```python
from pydantic_ai import ModelSettings

agent = Agent(
"openrouter:qwen/qwen3-coder:free",
system_prompt="you are a children story teller. Always max 3 sentences.",
model_settings=ModelSettings(temperature=0)
)


prompt = "A goldfish in  a aquarium..."

result = await agent.run(prompt)

print(result.output)
```

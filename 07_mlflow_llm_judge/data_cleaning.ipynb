## Clean email json file 

format that Mlflow requires:
- input -> inputs
- {"expected_response": {'sender_name': 'Erik Lindqvist', 'sender_email': 'erik.lindqvist@example.com', 'subject': 'Order Inquiry', 'body': 'I would like to know the status of my order.'}}


```python
df = df.rename({"input": "inputs"}, axis="columns")

df

```

```python
df["expectations"].apply(lambda row: {"expected:response": row})
```

```python
import json
df["expectations"].apply(lambda row: {"expected:response": json.dumps(row)})
df
```

```python
df.to_json("cleaned_emails.json", index=False, orient="records")
```
